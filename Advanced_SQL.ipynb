Q1. What is a Common Table Expression (CTE), and how does it improve SQL query readability?

A Common Table Expression (CTE) is a temporary result set defined using the WITH clause that exists only within the execution scope of a query. It improves readability by breaking complex subqueries or nested logic into modular, named steps that can be referenced like regular tables.

Q2. Why are some views updatable while others are read-only? Explain with an example.

Updatable Views: Map directly to a single table’s rows/columns without modifications, so SQL can map UPDATE statements straight back to the underlying base table.   
Example: A view selecting ProductID, ProductName, Price from Products.   

Read-Only Views: Contain aggregations (SUM, AVG), GROUP BY, DISTINCT, or multiple JOINs, making it impossible for the database engine to determine which base row to modify.   
Example: A view displaying Category and AVG(Price)

Q3. What advantages do stored procedures offer compared to writing raw SQL queries repeatedly?

Reusability & Maintainability: Encapsulates business logic once, preventing code duplication across applications.   

Performance: Pre-compiled and cached by the database execution engine, reducing query parsing time.   

Security: Prevents SQL injection and allows granting users execution access without directly exposing base tables.  

Q4. What is the purpose of triggers in a database? Mention one use case where a trigger is essential

A trigger is a database object that automatically executes SQL code in response to specific events (INSERT, UPDATE, or DELETE) on a table.  

 Essential Use Case: Audit Logging / Data Archiving—automatically recording changes or deleted rows into an audit/archive table whenever a row is modified or removed.

Q5. Explain the need for data modelling and normalization when designing a database.

Data Modelling: Defines the structure, relationships, and constraints between tables to accurately represent real-world entities.   

Normalization: Organizes database columns into distinct tables to reduce data redundancy, eliminate update/insert/delete anomalies, and maintain data integrity.

SELECT
    Category,
    COUNT(ProductID) AS TotalProducts,
    AVG(Price) AS AveragePrice

Q9. Create a stored procedure that accepts a category name and returns all products belonging to that
category.

```sql
DELIMITER //

CREATE PROCEDURE GetProductsByCategory(IN cat_name VARCHAR(50))
BEGIN
    SELECT ProductID, ProductName, Category, Price
    FROM Products
    WHERE Category = cat_name;
END //

DELIMITER ;
```[cite: 11, 12]

Q10. Create an AFTER DELETE trigger on the  
table
ProductArchive
timestamp.

```sql
-- Step 1: Create Archive Table
CREATE TABLE ProductArchive (
    ProductID INT,
    ProductName VARCHAR(100),
    Category VARCHAR(50),
    Price DECIMAL(10,2),
    DeletedAt TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);

-- Step 2: Create Trigger
DELIMITER //

CREATE TRIGGER trg_AfterDelete_Products
AFTER DELETE ON Products
FOR EACH ROW
BEGIN
    INSERT INTO ProductArchive (ProductID, ProductName, Category, Price, DeletedAt)
    VALUES (OLD.ProductID, OLD.ProductName, OLD.Category, OLD.Price, NOW());
END //

DELIMITER ;
```[cite: 11, 12]
